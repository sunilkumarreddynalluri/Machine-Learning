```python
#oops

import numpy as np
import pandas as pd
import sklearn
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix,accuracy_score,classification_report
from sklearn.naive_bayes import GaussianNB
import pickle
import json
import sys
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import sklearn
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix,accuracy_score,classification_report
import json
import sys

class Naives_Iris:
  def __init__(self,data):

    #reading data and splitting for train and test
    try:
      self.df = pd.read_csv(data)
      self.df = self.df.drop('Id',axis=1)
      self.df['Species'] = self.df['Species'].map({'Iris-setosa':0,'Iris-versicolor':1,'Iris-virginica':2}).astype(int)
      self.x = self.df.iloc[:,:-1]
      self.y = self.df.iloc[:,-1]
      self.x_train, self.x_test, self.y_train, self.y_test = train_test_split(
                  self.x, self.y, test_size=0.2, random_state=42)
    except Exception:
            err_ty, err_ms, err_line = sys.exc_info()
            print(f"Error is due to : {err_ty} from Line No : {err_line.tb_lineno} and msg is {err_ms}")

  def training(self):
    try:
      #Give Data to Naive Bayes
      from sklearn.naive_bayes import GaussianNB
      self.reg = GaussianNB()
      self.reg.fit(self.x_train,self.y_train)

      #Train Performance
      print('--------------Training Performance------------')
      train_prediction = self.reg.predict(self.x_train)
      print(confusion_matrix(self.y_train,train_prediction))
      print('========================================')
      print(accuracy_score(self.y_train,train_prediction))
      print('========================================')
      print(classification_report(self.y_train,train_prediction))
      print()

      #train_data to dictionary format
      acc = accuracy_score(self.y_train,train_prediction)
      conf_mat = confusion_matrix(self.y_train,train_prediction)
      train_report = classification_report(self.y_train,train_prediction,output_dict=True)
      train_data = {'Accuracy_Score':acc,'Confusion_matrix':conf_mat.tolist(),'Classification_report':train_report}

      #dump to json
      with open('train_Naives.json','w') as td:
        json.dump(train_data,td,indent=4)

    except Exception:
        err_ty, err_ms, err_line = sys.exc_info()
        print(f"Error is due to : {err_ty} from Line No : {err_line.tb_lineno} and msg is {err_ms}")

  def testing(self):
    try:

      #test performance
      print('--------------Testing Performance------------')
      test_prediction = self.reg.predict(self.x_test)
      print(confusion_matrix(self.y_test,test_prediction))
      print('========================================')
      print(accuracy_score(self.y_test,test_prediction))
      print('========================================')
      print(classification_report(self.y_test,test_prediction))
      print()

      #test_data to dictionary format
      accuracy = accuracy_score(self.y_test,test_prediction)
      confus_mat = confusion_matrix(self.y_test,test_prediction)
      test_report = classification_report(self.y_test,test_prediction,output_dict=True)
      test_data = {'Accuracy_Score':accuracy,'Confusion_matrix':confus_mat.tolist(),'Classification_report':test_report}

      #dump to json
      with open('test_Naives.json','w') as test:
        json.dump(test_data,test,indent=4)

    except Exception:
      err_ty, err_ms, err_line = sys.exc_info()
      print(f"Error is due to : {err_ty} from Line No : {err_line.tb_lineno} and msg is {err_ms}")

  def check(self):
    try:
      import warnings
      warnings.filterwarnings('ignore')

      r = self.reg.predict([[2.5,2.8,1.2,1.3]])[0]
      if r == 0:
        print(f'Flower is: Setosa')
      elif r == 1:
        print(f'Flower is: Versicolor')
      else:
        print(f'Flower is: Verginica')

      #creating pickle file
      import pickle
      with open('Naive Bayes.pkl','wb') as f:
        pickle.dump(self.reg,f)
    except Exception:
      err_ty, err_ms, err_line = sys.exc_info()
      print(f"Error is due to : {err_ty} from Line No : {err_line.tb_lineno} and msg is {err_ms}")

if __name__ == "__main__":
  obj = Naives_Iris('/content/Iris.csv')
  obj.training()
  obj.testing()
  obj.check()
```
