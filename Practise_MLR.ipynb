```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('/content/data.csv')
```

```python
df
```

```python
df = df.drop(columns=['date','street','statezip'])
```

```python
df
```

```python
df.shape
```

```python
df['city'].value_counts()
```

```python
a = {}
s = 0

for i in df['city']:
  if i not in a:
    a[i] = s
    s = s+1
  else:
    pass
df['city'] = df['city'].map(a).astype(int)
```

```python
df['city'].unique()
```

```python
df['country'].value_counts()
```

```python
df['country']=df['country'].map({'USA':0}).astype(int)
```

```python
df['country']
```

```python
x = df.iloc[:,1: ]
y = df.iloc[:,0]
```

```python
x
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(x,y,test_size = 0.2,random_state=15)
```

```python
len(X_train),len(y_train)
```

```python
len(X_test),len(y_test)
```

```python
from sklearn.linear_model import LinearRegression
reg = LinearRegression()
reg.fit(X_train,y_train)
```

```python
reg.coef_
```

```python
reg.intercept_
```

```python
#train
```

```python
y_train_predictions = reg.predict(X_train)
```

```python
from sklearn.metrics import root_mean_squared_error,r2_score
print(f'loss:{root_mean_squared_error(y_train,y_train_predictions)}')
print(f'accuracy:{r2_score(y_train,y_train_predictions)}')
```

```python
#test
```

```python
y_test_predictions = reg.predict(X_test)
```

```python
print(f'loss:{root_mean_squared_error(y_test,y_test_predictions)}')
print(f'accuracy:{r2_score(y_test,y_test_predictions)}')
```
