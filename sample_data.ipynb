```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('/content/Simple linear regression.csv')


```

```python
df
```

```python
x = df['SAT']
y = df['GPA']
```

```python
import sklearn
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=5)
```

```python
len(x_train) , len(y_train)
```

```python
len(x_test) , len(x_test)
```

```python

training_data = pd.DataFrame({'x_train':x_train,
                              'y_train':y_train}
                             )

training_data
```

```python
testing_data = pd.DataFrame({'x_test':x_test,
                             'y_test':y_test}
                            )

testing_data
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(5,3))
plt.title('simple linear regression')
plt.xlabel('SAT')
plt.ylabel('GPA')
plt.scatter(x = x_train, y = y_train, color= 'blue', marker = 'd')
plt.show
```

```python
x_train = x_train.values.reshape(-1,1)
```

```python
x_train.ndim
```

```python
from sklearn.linear_model import LinearRegression
reg = LinearRegression()
reg.fit(x_train , y_train)
```

```python
reg.coef_
```

```python
reg.intercept_
```

```python
sunil_model = reg.predict(x_train)
```

```python
sunil_model
```

```python
training_data['sunil_model'] = sunil_model
```

```python
training_data
```

```python
#mean square error
a = 0
for i in training_data.index:
   a  = a + (training_data['y_train'][i] - training_data['sunil_model'][i])**2

print(f'Loss value from MSE : {a / len(training_data)}')
```

```python
from sklearn.metrics import mean_squared_error

mean_squared_error(y_train , sunil_model)
```

```python
#root mean square error
a = 0
for i in training_data.index:
   a  = a + (training_data['y_train'][i] - training_data['sunil_model'][i])**2

print(f'Loss value from RMSE : { np.sqrt(a / len(training_data))}')
```

```python
from sklearn.metrics import mean_squared_error
mean_squared_error(y_train , sunil_model)
```

```python
#train accuracy
from sklearn.metrics import r2_score
r2_score(y_train , sunil_model)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(5,3))
plt.title('simple linear regression')
plt.xlabel('SAT')
plt.ylabel('GPA')
plt.scatter(x = x_train, y = sunil_model, color= 'blue', marker = 'd')
plt.show
```
