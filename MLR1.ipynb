```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import sklearn
```

```python
df = pd .read_csv('/content/Iris.csv')
df
```

```python
df.info()
```

```python
df = df.drop(['Id'],axis=1)
```

```python
df.shape
```

```python
df['Species'].value_counts()
```

```python
df['Species'].unique()
```

```python
df.info()
```

```python
df['Species'] = df['Species'].map({'Iris-setosa':0,'Iris-versicolor':1,'Iris-virginica':2}).astype(int)
```

```python
df
```

```python
df.info()
```

```python
x = df.iloc[ : , :-1]
x
```

```python
y = df.iloc[ : , -1]
y
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=22)
```

```python
len(X_train),len(y_train)
```

```python
len(X_test),len(y_test)
```

```python
from sklearn.linear_model import LinearRegression
reg = LinearRegression()
reg.fit(X_train,y_train)
```

```python
reg.coef_
```

```python
reg.intercept_
```

```python
#train performance

y_train_prediction = reg.predict(X_train)
```

```python
from sklearn.metrics import root_mean_squared_error,r2_score
```

```python
print(f'loss:{root_mean_squared_error(y_train,y_train_prediction)}')
print(f'accuracy:{r2_score(y_train,y_train_prediction)}')
```

```python
# test
y_test_prediction = reg.predict(X_test)
```

```python
print(f'loss:{root_mean_squared_error(y_test,y_test_prediction)}')
print(f'accuracy:{r2_score(y_test,y_test_prediction)}')
```
