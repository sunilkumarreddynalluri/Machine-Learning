```python
#oops


import sys
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import root_mean_squared_error,r2_score
```

```python

class MLR:
  def __init__(self,data):
    try:
      self.df = pd.read_csv(data)
      self.df['Position'] = self. df['Position'].map({'Business Analyst':0,'Junior Consultant':1,'Senior Consultant':2,'Manager':3,
                                     'Country Manager':4,'Region Manager':5,'Partner':6,
                                     'Senior Partner':7,'C-level':8,'CEO':9}).astype(int)
      self.x = self.df.iloc[ : , : -1]
      self.y = self.df.iloc[ : ,-1]
      self.X_train, self.X_test, self.y_train, self.y_test = train_test_split(self.x, self.y, test_size=0.2, random_state=42)

    except Exception as e:
      err_ty,err_ms,err_line = sys.exc_info()
      print(f'Error is due to : {err_ty} from Line No : {err_line.tb_lineno}')

  def training(self):
    try:
      self.reg = LinearRegression()
      self.reg.fit(self.X_train,self.y_train)
      self.y_train_predictions = self.reg.predict(self.X_train)
      print(f'Train Loss : {root_mean_squared_error(self.y_train , self.y_train_predictions)}')
      print(f'Train Accuracy : {r2_score(self.y_train , self.y_train_predictions)}')
    except Exception as e:
      err_ty,err_ms,err_line = sys.exc_info()
      print(f'Error is due to : {err_ty} from Line No : {err_line.tb_lineno}')

  def testing(self):
    try:
      self.y_test_predictions = self.reg.predict(self.X_test)
      print(f'Test Loss : {root_mean_squared_error(self.y_test , self.y_test_predictions)}')
      print(f'Test Accuracy : {r2_score(self.y_test , self.y_test_predictions)}')
    except Exception as e:
      err_ty,err_ms,err_line = sys.exc_info()
      print(f'Error is due to : {err_ty} from Line No : {err_line.tb_lineno}')

  def check(self):
    try:
      Position = 5
      Level = 5
      print(f'Total salary : {self.reg.predict([[Position,Level]])[0]}')
    except Exception as e:
      err_ty,err_ms,err_line = sys.exc_info()
      print(f'Error is due to : {err_ty} from Line No : {err_line.tb_lineno}')

if __name__ == "__main__":
  obj = MLR('/content/Position_Salaries.csv')
  obj.training()
  obj.testing()
  obj.check()
```
